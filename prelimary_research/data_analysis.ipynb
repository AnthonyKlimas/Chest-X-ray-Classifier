````python
%pip install pandas matplotlib seaborn pillow

"""
NIH Chest X-ray 14 Metadata Inspection
"""

from pathlib import Path
import pandas as pd

csv_candidates = [
    Path("Data_Entry_2017_v2020.csv"),
    Path("./Data_Entry_2017_v2020.csv"),
    Path("../Data_Entry_2017_v2020.csv"),
]

csv_path = next((p for p in csv_candidates if p.exists()), None)
if csv_path is None:
    raise FileNotFoundError("Could not find Data_Entry_2017_v2020.csv in expected locations.")

data_frame = pd.read_csv(csv_path)

# Convert numeric columns and rename resolution fields for easier plotting.
data_frame["Patient Age"] = pd.to_numeric(data_frame["Patient Age"], errors="coerce")
data_frame["Follow-up #"] = pd.to_numeric(data_frame["Follow-up #"], errors="coerce")
data_frame["Image Width"] = pd.to_numeric(data_frame["OriginalImage[Width"], errors="coerce")
data_frame["Image Height"] = pd.to_numeric(data_frame["Height]"], errors="coerce")

data_frame.head()
````

````python
"""
NIH Chest X-ray 14 Metadata Distribution
Author: Nicholas J. Calabro 
"""

import matplotlib.pyplot as plt
from collections import Counter


# Find all labels
all_labels = []
for labels in data_frame['Finding Labels']:
    all_labels.extend([label.strip() for label in labels.split('|')])

# Count each label
label_counts = Counter(all_labels)

# Plot bar chart
plt.figure(figsize=(12,6))
plt.bar(label_counts.keys(), label_counts.values(), color='skyblue')
plt.title('NIH Chest X-ray 14: Finding Labels Distribution')
plt.xlabel('Diagnosis')
plt.ylabel('Dataset Frequency Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
````

````python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes = axes.flatten()

# Histogram: patient ages
sns.histplot(data_frame["Patient Age"].dropna(), bins=30, kde=True, ax=axes[0], color="#2a9d8f")
axes[0].set_title("Patient Age Distribution")

# Histogram: follow-up counts
sns.histplot(data_frame["Follow-up #"].dropna(), bins=30, ax=axes[1], color="#e76f51")
axes[1].set_title("Follow-up Number Distribution")

# Boxplot: age by sex
sns.boxplot(data=data_frame, x="Patient Sex", y="Patient Age", ax=axes[2], color="#8ecae6")
axes[2].set_title("Patient Age by Sex")

# Correlation heatmap for numeric metadata
numeric_cols = ["Patient Age", "Follow-up #", "Image Width", "Image Height"]
corr = data_frame[numeric_cols].corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="YlGnBu", fmt=".2f", ax=axes[3])
axes[3].set_title("Correlation Heatmap (Metadata)")

# Count plot: view position
view_counts = data_frame["View Position"].value_counts().head(10)
sns.barplot(x=view_counts.index, y=view_counts.values, ax=axes[4], color="#264653")
axes[4].set_title("View Position Distribution")
axes[4].set_xlabel("View Position")
axes[4].set_ylabel("Count")

# Label count plot (top 10 findings)
all_labels = data_frame["Finding Labels"].str.split("|").explode().str.strip()
top_labels = all_labels.value_counts().head(10)
sns.barplot(x=top_labels.values, y=top_labels.index, ax=axes[5], color="#f4a261")
axes[5].set_title("Top 10 Finding Labels")
axes[5].set_xlabel("Count")
axes[5].set_ylabel("Label")

for ax in axes:
    ax.tick_params(axis="x", rotation=30)

plt.tight_layout()
plt.show()
````

### Written interpretations

##### Class Imbalance, Bias and Noise
There is a strong class imbalance. The "No finding" label appears more than other conditions, while diseases such as Hernia and pneumonia have very few samples compared to the rest. In addition, the analysis shows that negative cases are more common than positive cases for each disease. This imbalance can bias any future model toward the majority class and reduce its ability to detect rare conditions.

There is a potential bias in the dataset. The distribution of patient age is not uniform, and the differences between male and female patients are observed. The presence of different positions (PA and AP) means that images are captured under different conditions, which may influence the model learning.

Noise is another concern. The sample images show variation in brightness, positioning, and quality. Some images also include artifacts such as text labels (L or R), borders, or medical equipment. In addition, the metadata shows variability in image dimensions and aspect ratios, meaning that images are not consistent across the dataset. The multi-label nature of the dataset increases complexity, since multiple diseases can appear in a single image, making classification more challenging.

### Preliminary Plan for Preprocessing
The class imbalance must be handled. Techniques such as class weighting, resampling, or data augmentation can be applied to be sure that rare diseases are represented during training.

The dataset includes multi-label entries, meaning one image can have multiple diseases. This requires transforming the labels into a proper multi-label format. it might be necessary to separate combined labels for consistency across all categories.

Image standardization is required. Due to the variability in image dimensions, aspect ratios, and quality, all images should be resized to a consistent resolution and normalized. This step is importanto for the model to receive uniform input.

The effect of different view positions (PA vs AP) should be considered. This feature may need to be encoded or incorporated, as it can introduce bias into the model.

Noise reduction is also important. Some images contain artifacts such as text labels, borders, or medical equipment, which may need to be removed or reduced to prevent the model from learning irrelevant patterns.
